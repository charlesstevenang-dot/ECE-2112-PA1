A. WORD ROTATION PROBLEM

```python
def rotate_word(word: str) -> str:
 
    if not isinstance(word, str) or not word.strip():
        raise ValueError("Argument must be a non-empty string.")

    if len(word)<=1:
        return word

    return word[-1] + word[1:-1] + word[0]
    
```

Examples of the function "process_text()":

```python
print(rotate_word("Python"))
```

```python
print(rotate_word("G"))
```

B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name: str, last_name: str) -> str:
    
    final_first = first_name.lower().replace(" ", "")

    final_last = last_name.lower().replace(" ", "")

    return final_first + "." + final_last
    

    
```

Examples of the function "make_username()":

```python
make_username("Charles Steven" , "Ang")
```

```python
make_username("James Harold", "Batterson")
```

C. BOOKEND SWAP PROBLEM

```python
def swap_bookend(items: list) -> list:

    first, *middle, last = items

    return last, *middle, first
    
```

Examples of the function "swap_bookend":

```python
swap_bookend([1,2,3,4,5])
```

```python
swap_bookend(["red", "orange", "yellow", "green", "blue", "violet"])
```
